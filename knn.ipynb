```python

class KNN:
    def __init__(self, n_neighbors=5):
        self.n_neighbors = n_neighbors
        
    def euclidean_distance(self, x1, x2):
        return np.linalg.norm(x1 - x2)#x1 is testing data point and x2 is training datapoint and it computes distance
    
    def fit(self, X_train, y_train):#training phase mainly it stores trained data
        self.X_train = X_train
        self.y_train = y_train
    
    def predict(self, X):
         predictions = []
         for x in X:
             prediction = self._predict(x)
             predictions.append(prediction)
         return np.array(predictions)  
    
    def _predict(self, x):
        distances = []
        for x_train in self.X_train:
            distance = self.euclidean_distance(x, x_train)
            distances.append(distance)
        distances = np.array(distances)

        n_neighbors_idxs = np.argsort(distances)[: self.n_neighbors]
        labels = self.y_train[n_neighbors_idxs]                  
        labels = list(labels)
        most_occuring_value = max(labels, key=labels.count)
        return most_occuring_value
#Test point →Distance to all training points →Select K nearest →Majority vote →Prediction
```

```python
def train_test_split(X, y, random_state=42, test_size=0.2):
    n_samples = X.shape[0]#to get no.of rows
    np.random.seed(random_state)
    shuffled_indices = np.random.permutation(np.arange(n_samples))
    test_size = int(n_samples * test_size)
    test_indices = shuffled_indices[:test_size]
    train_indices = shuffled_indices[test_size:]
    X_train, X_test = X[train_indices], X[test_indices]
    y_train, y_test = y[train_indices], y[test_indices]

    return X_train, X_test, y_train, y_test
print("ty")
```

```python
# Train-test split (your own function)
X_train, X_test, y_train, y_test = train_test_split(
    X.values, y.values, test_size=0.2, random_state=42
)

# Scaling (from scratch)
mean = X_train.mean(axis=0)
std = X_train.std(axis=0)
std[std == 0] = 1

X_train = (X_train - mean) / std
X_test = (X_test - mean) / std

# Small data (for speed)
X_train_small = X_train[:200]
y_train_small = y_train[:200]
X_test_small = X_test[:50]
y_test_small = y_test[:50]

# Custom KNN
model = KNN(7)
model.fit(X_train_small, y_train_small)
  

```

```python
def compute_accuracy(y_true, y_pred):
    y_true = y_true.flatten()
    total_samples = len(y_true)
    correct_predictions = np.sum(y_true == y_pred)
    return (correct_predictions / total_samples)
print("yes")   
```

```python
predictions = model.predict(X_test_small)
accuracy = compute_accuracy(y_test_small, predictions)

print(f"Our model got accuracy score of: {accuracy}")
```

```python
sk_predictions = skmodel.predict(X_test)
sk_accuracy = compute_accuracy(y_test, sk_predictions)
print(f" sklearn-model got accuracy score of : {sk_accuracy}")
```
